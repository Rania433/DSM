```python
import numpy as np
x = np.array([[1,2],[4,5]])
U,S,VT = np.linalg.svd(x)
print("U =\n",U)

print("\nSingular values =",S)

print("\nVT =",VT)

sigma = np.diag(S)
rec=U @ sigma @ VT
print("reconstructed matrix:",rec)

```
